```python
import pandas as pd
from pathlib import Path

from data.extract_github_repo import build_repo_dataframe
from features.build_features import build_all_features


def build_feature_df_from_repos(repos, output_dir, merged_filename, failed_filename):
    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)

    feature_list = []
    failed_repos = []

    for repo in repos:
        print(f"\nProcessing: {repo}")

        try:
            df2 = build_repo_dataframe(repo)
            features = build_all_features(df2)

            features.insert(0, "repo_name", repo)

            safe_repo_name = repo.replace("/", "__")
            repo_output_path = output_dir / f"{safe_repo_name}_features.csv"

            features.to_csv(
                repo_output_path,
                index=False,
                encoding="utf-8-sig"
            )

            feature_list.append(features)

            print(f"Saved: {repo_output_path}")
            print("Feature shape:", features.shape)

        except Exception as e:
            failed_repos.append({
                "repo_name": repo,
                "error": str(e)
            })

            print(f"Failed: {repo}")
            print(e)

    if feature_list:
        merged_df = pd.concat(feature_list, ignore_index=True)

        all_output_path = output_dir / merged_filename
        merged_df.to_csv(
            all_output_path,
            index=False,
            encoding="utf-8-sig"
        )

        print("\nMerged feature shape:", merged_df.shape)
        print(f"Saved merged file: {all_output_path}")

    else:
        merged_df = pd.DataFrame()
        print("No features created.")

    if failed_repos:
        failed_df = pd.DataFrame(failed_repos)

        failed_output_path = output_dir / failed_filename
        failed_df.to_csv(
            failed_output_path,
            index=False,
            encoding="utf-8-sig"
        )

        print(f"Saved failed repo list: {failed_output_path}")

    return merged_df
```

```python
import pandas as pd
from pathlib import Path

from data.extract_github_repo import build_repo_dataframe
from features.build_features import build_all_features


def resume_build_feature_df_from_repos(
    repos,
    output_dir,
    merged_filename,
    failed_filename
):
    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)

    feature_list = []
    failed_repos = []

    for repo in repos:
        print(f"\nProcessing: {repo}")

        safe_repo_name = repo.replace("/", "__")
        repo_output_path = output_dir / f"{safe_repo_name}_features.csv"

        # 이미 저장된 파일은 건너뛰기
        if repo_output_path.exists():
            print(f"Skip already exists: {repo_output_path}")
            continue

        try:
            df2 = build_repo_dataframe(repo)
            features = build_all_features(df2)

            features.insert(0, "repo_name", repo)

            features.to_csv(
                repo_output_path,
                index=False,
                encoding="utf-8-sig"
            )

            feature_list.append(features)

            print(f"Saved: {repo_output_path}")
            print("Feature shape:", features.shape)

        except Exception as e:
            failed_repos.append({
                "repo_name": repo,
                "error": str(e)
            })

            print(f"Failed: {repo}")
            print(e)

    # 이번에 새로 만든 것만 merge
    if feature_list:
        new_merged_df = pd.concat(feature_list, ignore_index=True)

        new_output_path = output_dir / merged_filename
        new_merged_df.to_csv(
            new_output_path,
            index=False,
            encoding="utf-8-sig"
        )

        print("\nNew merged feature shape:", new_merged_df.shape)
        print(f"Saved new merged file: {new_output_path}")

    else:
        new_merged_df = pd.DataFrame()
        print("No new features created.")

    if failed_repos:
        failed_df = pd.DataFrame(failed_repos)

        failed_output_path = output_dir / failed_filename
        failed_df.to_csv(
            failed_output_path,
            index=False,
            encoding="utf-8-sig"
        )

        print(f"Saved failed repo list: {failed_output_path}")

    return new_merged_df
```

```python
importance_df = pd.DataFrame({
    "feature": X.columns,
    "importance": model.feature_importances_
})

importance_df = importance_df.sort_values(
    by="importance",
    ascending=False
)

importance_df.head(20)
```
