```python
import pandas as pd
repos = [
    "pandas-dev/pandas",
    "numpy/numpy",
    "scikit-learn/scikit-learn",
    "matplotlib/matplotlib",
    "psf/requests",
    "fastapi/fastapi",
    "django/django",
    "pallets/flask",
]

feature_list = []

for repo in repos:
    print(f"Processing: {repo}")

    try:
        df2 = build_repo_dataframe(repo)
        features = build_all_features(df2)

        features.insert(0, "repo_name", repo)
        feature_list.append(features)

        print(repo, features.shape)

    except Exception as e:
        print(f"Failed: {repo}")
        print(e)

all_repo_features = pd.concat(feature_list, ignore_index=True)

all_repo_features.shape
```
